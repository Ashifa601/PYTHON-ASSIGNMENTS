```python
#NUMPY
import numpy as np

# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)

```

```python
# 2. Inspection and Properties
print("\nShape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# 3. Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("\nTemperatures in Fahrenheit:")
print(temperatures_f)

# Maximum, Minimum and Mean
print("\nMaximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", round(np.mean(temperatures_w1), 2))
```

```python
# 4. Array Slicing and Indexing
first_three_days = temperatures_w1[:3]
weekend = temperatures_w1[-2:]
middle_three_days = temperatures_w1[2:5]

print("\nFirst Three Days:", first_three_days)
print("Weekend:", weekend)
print("Middle Three Days:", middle_three_days)
```

```python
# 5. Create a 2D Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("\n2D Temperature Array:")
print(temperatures)
```

```python
# 6. Inspect the 2D Array
print("\n2D Array Shape:", temperatures.shape)
print("2D Array Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
# Extract each week
week1 = temperatures[0]
week2 = temperatures[1]

print("\nWeek 1:", week1)
print("Week 2:", week2)
```

```python

# Extract weekends for both weeks
weekends = temperatures[:, -2:]

print("\nWeekend Temperatures for Both Weeks:")
print(weekends)
```

```python
#PANDAS
import pandas as pd

# 1. Creating Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Marks Series:")
print(marks)
```

```python
# 2. Indexing and Slicing

# Mark of 1st rank student
print("\nMark of 1st rank student:")
print(marks.iloc[0])

# Top 3 ranks using loc
print("\nTop 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Mark of 3rd rank student using iloc
print("\nMark of 3rd rank student:")
print(marks.iloc[2])

# Marks greater than 90
print("\nRanks with marks greater than 90:")
print(marks[marks > 90])
```

```python
# 3. Manipulating Series

# Modify Rank1 mark
marks.loc['Rank1'] = 100

print("\nAfter modifying Rank1:")
print(marks)

# Remove Rank5
marks = marks.drop('Rank5')

print("\nAfter removing Rank5:")
print(marks)

# Calculate CGPA
cgpa = marks / 10

print("\nCGPA:")
print(cgpa)
```

```python
#PANDAS DATAFRAME

# 1. Creating Pandas DataFrame

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
```

```python
# 2. Data Exploration

# Display DataFrame
print("Transactions DataFrame:")
print(transactions)

# Head
print("\nFirst 5 rows:")
print(transactions.head())

# Tail
print("\nLast 5 rows:")
print(transactions.tail())


```

```python
# Shape
print("\nShape:")
print(transactions.shape)

# Column names
print("\nColumn Names:")
print(transactions.columns)

# Data types
print("\nData Types:")
print(transactions.dtypes)

# Basic information
print("\nBasic Information:")
transactions.info()
```

```python
# ProductCategory and Amount
print("\nProduct Category and Amount:")
print(transactions[['ProductCategory', 'Amount']])

# Last 3 columns
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])

```

```python
# Region = North AND Amount > 200
print("\nNorth Region with Amount greater than 200:")
print(
    transactions[
        (transactions['Region'] == 'North') &
        (transactions['Amount'] > 200)
    ]
)

# ProductCategory value counts
print("\nProduct Category Counts:")
print(transactions['ProductCategory'].value_counts())

# Unique Regions
print("\nUnique Regions:")
print(transactions['Region'].unique())

# Mean Amount by Region
print("\nMean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. Manipulating the DataFrame

# Modify Amount for TransactionID 102
transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

print("\nAfter modifying TransactionID 102:")
print(transactions)

```

```python
# Add Discount column
transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter adding Discount:")
print(transactions)
```

```python
# Remove TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

print("\nAfter removing TransactionID 109:")
print(transactions)
```

```python
# Delete Discount column
transactions = transactions.drop('Discount', axis=1)

print("\nFinal DataFrame:")
print(transactions)
```
